# Project Description

This project develops a survival prediction workflow for the CIBMTR post-HCT dataset. The training data is split into training, validation, and holdout test sets, then used to create survival-based regression targets and train multiple machine learning models.

The final prediction is produced with a rank-based weighted ensemble and evaluated using the stratified C-index across race groups.
