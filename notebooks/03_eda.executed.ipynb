# Exploratory Data Analysis

Plot generated target distributions after target creation.

```python
import sys
from pathlib import Path

project_root = Path.cwd().resolve()
if project_root.name == "notebooks":
    project_root = project_root.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
```

```python
from src.config import CFG
from src.eda import EDA
from src.modeling import MD
from src.preprocessing import get_categorical_columns, load_training_data, split_dataset
```

```python
dataframe = load_training_data(CFG.train_path)
train_data, validation_data, test_data = split_dataset(
    dataframe,
    validation_size=CFG.validation_size,
    test_size=CFG.test_size,
    random_state=CFG.random_state,
    stratify_column=CFG.split_stratify_column,
)
cat_cols = get_categorical_columns(train_data)
md = MD(CFG.color, train_data, cat_cols, CFG.early_stop, CFG.penalizer, CFG.n_splits, CFG.random_state)
train_data = md.create_targets()
eda = EDA(CFG.color, train_data)
```

## Target score analysis
The target-generation step produced very strong Kaplan-Meier and Nelson-Aalen training scores. These scores are expected to be optimistic because the targets are derived from survival estimators on the training split.


```python
eda.histogram_distribution_plot("cox_hazard", "Cox Target")
eda.histogram_distribution_plot("km_survival", "Kaplan-Meier Target")
eda.histogram_distribution_plot("na_hazard", "Nelson-Aalen Target")
eda.histogram_distribution_plot("event_time", "Target for Cox-Loss Models")
```

## Distribution analysis
The target distributions should be checked before model training. Skewed or compressed targets can make rank-based ensembling more stable than averaging raw predictions.
